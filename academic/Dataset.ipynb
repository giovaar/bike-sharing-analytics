### Dataset Day

*Cargar dataset*

```python
# Se intenta cargar el dataset usando coma como separador.
df = pd.read_csv("day.csv")

# Se muestran las primeras cinco filas para verificar que cargó correctamente.
df.head()
```

### Observaciones y variables

```python
# Se obtiene la cantidad de filas y columnas del dataset.
observaciones, variables = df.shape

# Se imprimen los resultados.
print("Cantidad de observaciones:", observaciones)
print("Cantidad de variables:", variables)
```

### Informacion general del dataset

```python
# Se muestra información general del dataset:
# nombres de columnas, cantidad de valores no nulos y tipos de datos.
df.info()
```

### Tipo de dato por variable

```python
# Se crea una tabla con los tipos de datos de cada variable.
tipos_datos = pd.DataFrame({
    "Variable": df.columns,
    "Tipo de dato": df.dtypes.values
})

# Se muestra la tabla.
tipos_datos
```

### Clasificacion de variables categoricas y numericas

```python
# Se identifican las variables numéricas.
variables_numericas = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Se identifican las variables categóricas o de texto.
variables_categoricas = df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

# Se muestran los resultados.
print("Variables numéricas:")
print(variables_numericas)

print("\nVariables categóricas:")
print(variables_categoricas)

print("\nCantidad de variables numéricas:", len(variables_numericas))
print("Cantidad de variables categóricas:", len(variables_categoricas))
```

### Estadisticas descriptivas

```python
# Se muestran estadísticas descriptivas para variables numéricas.
df.describe()
```

```python
# Se muestran estadísticas descriptivas para variables categóricas.
df.describe(include=["object", "category", "bool"])
```

### Analisis de datos faltantes

```python
# Se calcula la cantidad de valores faltantes por variable.
cantidad_faltantes = df.isnull().sum()

# Se calcula el porcentaje de valores faltantes por variable.
porcentaje_faltantes = (df.isnull().mean() * 100).round(2)

# Se crea una tabla resumen con valores faltantes, porcentaje y tipo de dato.
resumen_faltantes = pd.DataFrame({
    "Valores faltantes": cantidad_faltantes,
    "Porcentaje faltante (%)": porcentaje_faltantes,
    "Tipo de dato": df.dtypes
})

# Se ordena de mayor a menor cantidad de valores faltantes.
resumen_faltantes = resumen_faltantes.sort_values(by="Valores faltantes", ascending=False)

# Se muestra el resumen.
resumen_faltantes
```

### Variable con mayor ausencia

```python
# Se identifica la variable con mayor cantidad de valores faltantes.
variable_mayor_ausencia = resumen_faltantes.index[0]

# Se obtiene la cantidad de valores faltantes de esa variable.
mayor_cantidad_faltantes = resumen_faltantes.loc[variable_mayor_ausencia, "Valores faltantes"]

# Se obtiene el porcentaje de ausencia de esa variable.
mayor_porcentaje_faltante = resumen_faltantes.loc[variable_mayor_ausencia, "Porcentaje faltante (%)"]

# Se muestran los resultados.
print("Variable con mayor ausencia:", variable_mayor_ausencia)
print("Cantidad de valores faltantes:", mayor_cantidad_faltantes)
print("Porcentaje de ausencia:", mayor_porcentaje_faltante, "%")
```

### Variables con datos faltantes

```python
# Se filtran únicamente las variables que tienen al menos un valor faltante.
variables_con_faltantes = resumen_faltantes[resumen_faltantes["Valores faltantes"] > 0]

# Se muestra la tabla.
variables_con_faltantes
```

# Limpieza del dataset Day.csv

```python
# Se calcula el total de valores faltantes.
total_faltantes = df.isnull().sum().sum()

# Se calcula la cantidad de filas duplicadas.
filas_duplicadas = df.duplicated().sum()

# Se identifican columnas completamente vacías.
columnas_completamente_vacias = df.columns[df.isnull().all()].tolist()

# Se seleccionan columnas numéricas para revisar valores infinitos.
df_numerico = df.select_dtypes(include=[np.number])

# Se verifica si hay valores infinitos en columnas numéricas.
valores_infinitos = np.isinf(df_numerico).sum().sum()

# Se muestran los resultados.
print("Total de valores faltantes:", total_faltantes)
print("Filas duplicadas:", filas_duplicadas)
print("Columnas completamente vacías:", columnas_completamente_vacias)
print("Valores infinitos en columnas numéricas:", valores_infinitos)

# Se determina si el dataset está limpio según criterios básicos.
if total_faltantes == 0 and filas_duplicadas == 0 and len(columnas_completamente_vacias) == 0 and valores_infinitos == 0:
    print("\nResultado: El dataset está aparentemente limpio.")
else:
    print("\nResultado: El dataset NO está completamente limpio y requiere tratamiento.")
```

### *Resumen de limpieza*

```python
# Se crea una tabla resumen con los principales aspectos evaluados.
resumen_limpieza = pd.DataFrame({
    "Aspecto evaluado": [
        "Cantidad de observaciones",
        "Cantidad de variables",
        "Total de valores faltantes",
        "Filas duplicadas",
        "Columnas completamente vacías",
        "Valores infinitos en columnas numéricas",
    ],
    "Resultado": [
        observaciones,
        variables,
        total_faltantes,
        filas_duplicadas,
        len(columnas_completamente_vacias),
        valores_infinitos,
    ]
})

# Se muestra el resumen.
resumen_limpieza
```
